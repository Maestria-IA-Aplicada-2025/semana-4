# **Comparación de las librerías Pandas y Polars**
Estas librerías permiten operativizar conjuntos de datos de forma estructurada

```python
# Importar las librerías necesarias
import pandas as pd
import polars as pl
import matplotlib.pyplot as plt
```

```python
# Crear un DataFrame con Pandas
data_pandas = {'columna1': [1, 2, 3, 4, 5], 'columna2': [10, 20, 30, 40, 50]}
df_pandas = pd.DataFrame(data_pandas)
```

```python
#Presentar DataFrame
df_pandas
```

```python
# Crear un DataFrame con Polars
data_polars = {'columna1': [1, 2, 3, 4, 5], 'columna2': [10, 20, 30, 40, 50]}
df_polars = pl.DataFrame(data_polars)
```

```python
#Presentar Dataframe
df_polars
```

```python
# Análisis con Pandas
# Calcular la suma de cada columna
sum_pandas = df_pandas.sum()
```

```python
# Análisis con Polars
# Calcular la suma de cada columna
sum_polars = df_polars.select(pl.col('*').sum())
```

```python
# Mostrar los resultados
sum_pandas, sum_polars
```

```python
# Visualización con Pandas
df_pandas.plot(kind='bar', x='columna1', y='columna2', title="Gráfico de barras con Pandas")
plt.show()
```

```python
# Visualización con Polars utilizando la librería Altair
!pip install altair>=5.4.0
import altair as alt

alt.Chart(df_polars.to_pandas()).mark_bar().encode(
    x='columna1',
    y='columna2',
).properties(
    title="Gráfico de barras con Polars"
)
```

```python
# Visualización con Polars desde un Dataframe de Pandas
!pip install pyarrow
df_polars.to_pandas().plot(kind='bar', x='columna1', y='columna2', title="Gráfico de barras con Polars")
plt.show()
```
